In this workspace, you'll scrape the novel Moby Dick from the website [Project Gutenberg](https://www.gutenberg.org/) (which contains a large corpus of books) using the Python `requests` package. You'll extract words from this web data using `BeautifulSoup` before analyzing the distribution of words using the Natural Language ToolKit (`nltk`) and `Counter`.

The Data Science pipeline you'll build in this workspace can be used to visualize the word frequency distributions of any novel you can find on Project Gutenberg.

```python
# Import and download packages
import requests
from bs4 import BeautifulSoup
import nltk
from collections import Counter
nltk.download('stopwords')
```

```python
# Getting web requests
import requests 

r = requests.get("https://s3.amazonaws.com/assets.datacamp.com/production/project_147/datasets/2701-h.htm")
print(r.status_code)                  #request code 200 - Successful returned
```

```python
#Extracting HTML with BeautifulSoup 
from bs4 import BeautifulSoup

html = r.text                #getting html content
html_soup = BeautifulSoup(html, "html.parser")            #convert into nested html structure
moby_text = html_soup.get_text()              #getting text from the html style
#print(moby_test)
```

```python
import nltk

tokenizer = nltk.tokenize.RegexpTokenizer("\w+")      #setting up tokenizer to remove special characters
tokens = tokenizer.tokenize(moby_text)    #applying tokenizer on our text 
#print(tokens)
```

```python
words = [token.lower() for token in tokens]     #list comprehension loop to convert every word into lowercase
#print(lower_case)

from nltk.corpus import stopwords
nltk.download("stopwords")                   #downloading "stop words"
stop_words = stopwords.words("english")       #setting stop words from nltk and storing in stop_words        

words_no_stop = [word for word in words if word not in stop_words]    #append if token (word) not in unwanted words
#print(words_no_stop)

count = Counter(words_no_stop)                #applying counter to the list to count # of words
top_ten = count.most_common(10)               #count the # of words
print(f'top 10 most common words are {top_ten}')           #printing results
```
